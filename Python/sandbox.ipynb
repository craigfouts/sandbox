```python
import matplotlib.pyplot as plt
import numpy as np
import requests
```

```python
from bs4 import BeautifulSoup
from html.parser import HTMLParser
```

```python
# url = 'https://docs.google.com/document/d/e/2PACX-1vTMOmshQe8YvaRXi6gEPKKlsC6UpFJSMAk4mQjLm_u1gmHdVVTaeh7nBNFBRlui0sTZ-snGwZM4DBCT/pub'
url = 'https://docs.google.com/document/d/e/2PACX-1vSvM5gDlNvt7npYHhp_XfsJvuntUhq184By5xO_pA4b_gCWeXb6dM6ZxwN8rE6S4ghUsCj2VKR21oEP/pub'

def get_table(url):
    response = requests.get(url)

    if response.status_code == 200:
        soup = BeautifulSoup(response.text, 'html.parser')
        table = soup.find('table')

        return table
    
    print(f'Error: {response.status_code}')

def parse_row(row):
    cols = row.find_all('td')
    c = cols[1].get_text()
    x = cols[0].get_text()
    y = cols[2].get_text()

    return c, x, y

def parse_table(table):
    rows = table.find_all('tr')
    chars, x_coords, y_coords = [], [], []


    for row in rows[1:]:
        char, x_coord, y_coord = parse_row(row)
        chars.append(char)
        x_coords.append(int(x_coord))
        y_coords.append(int(y_coord))

    return chars, x_coords, y_coords

def get_y_chars(y_coord, chars, x_coords, y_coords):
    check_y_idx = lambda i: y_coords[i] == y_coord
    y_x_coords, y_chars, sorted_y_chars = [], [], []

    for idx in filter(check_y_idx, range(len(y_coords))):
        y_x_coords.append(x_coords[idx])
        y_chars.append(chars[idx])

    for idx in range(len(x_coords)):
        if idx in y_x_coords:
            sorted_y_chars.append(y_chars[y_x_coords.index(idx)])
        else:
            sorted_y_chars.append(' ')

    return sorted_y_chars

def parse_url(url):
    table = get_table(url)
    chars, x_coords, y_coords = parse_table(table)

    for y_coord in range(max(y_coords), -1, -1):
        y_chars = get_y_chars(y_coord, chars, x_coords, y_coords)
        print(''.join(y_chars))

parse_url(url)
```
